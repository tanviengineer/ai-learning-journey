```python
# text data cleaning using nltk
import nltk
import string
```

```python
# download required nltk resources
nltk.download("stopwords")
nltk.download("punkt")
nltk.download("punkt_tab")

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PosterStemmer
```

```python
# create a stemmer object
stemmer =PosterStemmer()
```

```python
# load english stop word
stop_words =set (stopwords.words('english'))
```

```python
# resd text from a file
with open("input.txt","r",encoding ="utf-8")as file:
  text=file.read()

print("Original text :")
print(text)

# step1 convert text to lowercase
text =text.lower()

# step 2 remove punctuation
text =text.translste(str.maketrans('','',string.punctuation))

# step3 tokenize the text into words
words =word_tokenize(text)

# step4 remove stop word
cleaned_words =[]
for word in words:
  if word not in stop_words:
    cleaned_words.append(word)

# step5 perform stemming
stemmed_words =[]

for word in cleaned_words:
  stemmed_word=stemmer.stem(word)
  stemmed_words.aappned(stemmed_word)

# join word to from cleande text
cleaned_text =" ".join(stemmed_words)

print('\n Cleaned text :')
print(cleaned_text)

# save cleaned text to a new file
with open("cleaned_output.txt","w",encoding="utf-8")as filr:
  file.write(cleaned_text)
```
