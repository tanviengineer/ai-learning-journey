# **Z Test **

```python
import numpy as np
from scipy.stats import norm
```

```python
sample =[172,174,168,169,171,173,175,170,169,172]
pop_mean = 170
pop_std = 3
sample_mean= np.mean(sample)
n =len(sample)
```

```python
z_score= (sample_mean -pop_mean)/(pop_std/np.sqrt(n))
```

```python
z_score
```

```python
p_value =2*(1-norm.cdf(z_score))
```

```python
p_value
```

```python
alpha=0.05
if p_value<alpha:
  print("i will reh=ject the null hypothesis")
else:
  print("i will accept the null hypothesis")
```

# T **TEST**

```python
import numpy as np
from scipy import stats
```

```python
sample =[172,174,168,169,171,173,175,170,169,172]
sample_mean =np.mean(sample)
sample_std= np.std(sample,ddof =1)
n= len(sample)
mean_population = 170
```

```python
t_stats = (sample_mean - mean_population)/(sample_std- np.sqrt(n))
t_stats
```

```python
p_value =2*(1-stats.t.cdf(abs(t_stats),df=n-1))
alpha = 0.5
```

```python
if p_value<alpha:
  print("i will reject the null hypothesis")
else:
  print("i will accept the null hypothesis")
```

# **TWO SAMPLE TEST**

```python
import numpy as np
from scipy import stats
```

```python
group_A =[85,88,90,92,87,85,89,91,86,88]
group_B =[82,84,80,83,81,79,78,85,84,83]
```

```python
t_stats ,p_value =stats.ttest_ind(group_A,group_B,equal_var=False)
```

```python
t_stats
```

```python
alpha =0.05
```

```python
if p_value < alpha:
  print("i will reject the null hypotheis")
else:
  print("i will acecpt the null hypotheis")
```

# **ChiSquare Test**

```python

df = sns.load_dataset('titanic')
```

```python
df
```

```python
contingency_table = pd.crosstab(df['sex'],df['survived'])
```

```python
contingency_table
```

```python
expected
```

```python
if p_value < alpha:
    print("we reject the null hypothesis and there is a significant relationship between gender and survivl")
else:
    print("there is no connenction")
```

# **ANNOVA_test**

```python
import seaborn as sns
import pandas as pd
from scipy.stats import f_oneway
```

```python

df = sns.load_dataset('titanic')
```

```python
df
```

```python
df = df[['age','pclass']].dropna()
```

```python
df['pclass'].unique()
```

```python
class_1 = df[df['pclass'] == 1]['age']
class_2 = df[df['pclass'] == 2]['age']
class_3 = df[df['pclass'] == 3]['age']
class_1
```

```python
f_stats, p_value = f_oneway(class_1,class_2,class_3)
```

```python
f_stats
```

```python
p_value
```

```python
alpha = 0.05
```

```python
if p_value < alpha:
    print("reject the null hypothesis and there is a signafacant difference in atleat one passenger class")

else:
    print("there is no significant difference ")
```
