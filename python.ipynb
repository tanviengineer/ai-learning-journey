```python
#  used if and error happens --do not crash --go to except block

# try --code that may cause and error

# except --code that runs if an error occurs
```

```python
try:

   num = int (input("ENter a number : "))
   print(num)
except:
  print("ENter a valid number")
```

```python
try:

   num1 = int (input("ENter a number : "))
   num2 = int (input("ENter a number : "))
   print(num1+num2)
except:
  print("something went wrong. f")
```
