LIST

```python
fruits = [1,2,3,5,6,7,8,90,9,5,3,4,5]
print(fruits)
```

```python
# indexing
print(fruits[0])
print(fruits[2])
print(fruits[-4])
```

```python
# slicing
print(fruits[1:6])
print(fruits[4:7])
```

```python
# methods ()

# append()
fruits.append(1000)
print(fruits)
# remove()
fruits.remove(1)
print(fruits)
# sort()
fruits.sort()
print(fruits)
# decending sort()
fruits.sort(reverse= True)
print(fruits)

# lengh()
print(len(fruits))
# max()
print(max(fruits))
# min()
print(min(fruits))
# sum()
print(sum(fruits))
```

```python
# practice  ---find average
avg =sum(fruits)/len(fruits)
print(avg)
```
